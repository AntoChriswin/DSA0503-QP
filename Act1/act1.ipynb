```python
import sqlite3
```

```python
conn = sqlite3.connect("college.db")
cursor = conn.cursor()
```

```python
cursor.execute("PRAGMA foreign_keys = ON")
```

1. Create the Department Table (Parent Table) with attributes Department_ID
INTEGER PRIMARY KEY, Department_Name TEXT

```python
cursor.execute("create table department(department_id integer primary key, department_name text)")
```

2. Create the Student Table (Child Table) with the following attributes:
Student_ID INTEGER PRIMARY KEY,
Name TEXT,
Age INTEGER,
Department_ID INTEGER,
CGPA REAL,
FOREIGN KEY (Department_ID) REFERENCES Department(Department_ID)

```python
cursor.execute("create table student(student_id integer primary key,student name text, department_id integer , CGPA real ,"
" foreign key(department_id) references department(department_id))")
```

3.Display students from CSE department.

```python
cursor.execute("Select * from student  Join department on student.department_id = department.department_id where department.department_name = 'CSE'");
row = cursor.fetchall()
for i in row:
    print(i)
```

4.Count students in each department.

```python
cursor.execute("Select department.department_name,count(student.student_id) from department,student where department.department_id = student.department_id group by department.department_name");
row = cursor.fetchall()
for i in row:
    print(i)    
```

6. Delete a department

```python
cursor.execute("""DELETE FROM Department WHERE Department_ID = 3 """)
```

A foreign key is used to maintain the relationship between two tables. You cannot delete a parent record if it is still being referenced by a child table, because it would leave invalid or orphaned records.

7. Display the student’s name, student’s department and CGPA if the student’s CGPA is
greater than 8.

```python
cursor.execute("Select * from student join department on student.department_id = department.department_id where student.CGPA > 8");
row = cursor.fetchall()
for i in row:
    print(i)
```
