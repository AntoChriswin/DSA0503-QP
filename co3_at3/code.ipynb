```python
import pandas as pd

customer = pd.DataFrame({
    "Customer_ID": ["C101", "C102", "C103", "C104", "C105"],
    "Customer_Name": ["Arun", "Priya", "Rahul", "Sneha", "Karthik"],
    "City": ["Chennai", "Bangalore", "Hyderabad", "Mumbai", "Chennai"]
})


orders = pd.DataFrame({
    "Order_ID": ["O001", "O002", "O003", "O004", "O005",
                 "O006", "O007", "O008", "O009", "O010"],
    "Customer_ID": ["C101", "C102", "C101", "C103", "C104",
                    "C102", "C105", "C103", "C101", "C105"],
    "Order_Date": ["2026-08-01", "2026-08-02", "2026-08-03",
                   "2026-08-04", "2026-08-05", "2026-08-06",
                   "2026-08-07", "2026-08-08", "2026-08-09",
                   "2026-08-10"],
    "Order_Amount": [2500, 1800, 3200, 4500, 2100,
                     2700, 5000, 3500, 1500, 4200]
})

merged = pd.merge(customer, orders, on="Customer_ID")

print("Merged Data:")
print(merged[["Customer_Name", "City", "Order_Date", "Order_Amount"]])

total = merged.groupby("Customer_Name")["Order_Amount"].sum()

highest = total.idxmax()

print("\nCustomer with highest total order amount:")
print(highest)
print("Total Amount:", total.max())
```

```python
import pandas as pd

employee = pd.DataFrame({
    "Employee_ID": ["E101", "E102", "E103", "E104", "E105"],
    "Employee_Name": ["Arun", "Priya", "Rahul", "Sneha", "Karthik"],
    "Department": ["IT", "HR", "Finance", "IT", "Sales"]
})

attendance = pd.DataFrame({
    "Employee_ID": [
        "E101", "E101", "E101",
        "E102", "E102", "E102",
        "E103", "E103", "E103",
        "E104", "E104", "E104",
        "E105", "E105", "E105"
    ],
    "Date": [
        "2026-08-01", "2026-08-02", "2026-08-03",
        "2026-08-01", "2026-08-02", "2026-08-03",
        "2026-08-01", "2026-08-02", "2026-08-03",
        "2026-08-01", "2026-08-02", "2026-08-03",
        "2026-08-01", "2026-08-02", "2026-08-03"
    ],
    "Status": [
        "Present", "Present", "Absent",
        "Present", "Absent", "Present",
        "Absent", "Present", "Present",
        "Present", "Present", "Absent",
        "Present", "Absent", "Absent"
    ]
})


merged = pd.merge(employee, attendance, on="Employee_ID")

result = merged.groupby(
    ["Employee_ID", "Employee_Name", "Department", "Status"]
).size().unstack(fill_value=0)

result = result.rename(columns={
    "Present": "Present_Days",
    "Absent": "Absent_Days"
})

print("Attendance Summary:")
print(result.reset_index())
```

```python
import pandas as pd

students = pd.DataFrame({
    "Student_ID": ["S101", "S102", "S103", "S104", "S105"],
    "Student_Name": ["Arun", "Priya", "Rahul", "Sneha", "Karthik"],
    "Department": ["CSE", "IT", "ECE", "CSE", "IT"]
})

enrollment = pd.DataFrame({
    "Student_ID": [
        "S101", "S101",
        "S102", "S102",
        "S103", "S103",
        "S104", "S104",
        "S105", "S105"
    ],
    "Course_ID": [
        "C101", "C102",
        "C103", "C104",
        "C105", "C106",
        "C101", "C107",
        "C103", "C108"
    ],
    "Course_Name": [
        "Python", "Database",
        "Java", "Data Science",
        "Computer Networks", "Operating Systems",
        "Python", "Machine Learning",
        "Java", "Cloud Computing"
    ],
    "Credits": [4, 3, 4, 3, 4, 3, 4, 4, 4, 3]
})

merged = pd.merge(students, enrollment, on="Student_ID")

print("Merged Data:")
print(merged[[
    "Student_Name",
    "Department",
    "Course_Name",
    "Credits"
]])

total_credits = merged.groupby(
    ["Student_ID", "Student_Name"]
)["Credits"].sum()

print("\nTotal Credits Registered by Each Student:")
print(total_credits)
```

```python
import pandas as pd
sales = pd.DataFrame({
    "Sale_ID": ["S001", "S002", "S003", "S004", "S005",
                "S006", "S007", "S008", "S009", "S010"],
    "Product_ID": ["P101", "P102", "P103", "P104", "P105",
                   "P101", "P102", "P103", "P104", "P105"],
    "Region_ID": ["R01", "R02", "R01", "R03", "R02",
                  "R03", "R01", "R02", "R03", "R01"],
    "Quantity": [10, 8, 5, 12, 10, 7, 6, 9, 8, 4],
    "Amount": [5000, 4000, 3000, 6000, 5500,
               3500, 2800, 4500, 4200, 2200]
})
regions = pd.DataFrame({
    "Region_ID": ["R01", "R02", "R03"],
    "Region_Name": ["North", "South", "West"],
    "Manager": ["Arun", "Priya", "Rahul"]
})
merged = pd.merge(sales, regions, on="Region_ID")
print("Merged Data:")
print(merged[[
    "Region_Name",
    "Manager",
    "Product_ID",
    "Quantity",
    "Amount"
]])
total_sales = merged.groupby("Region_Name")["Amount"].sum()

print("\nTotal Sales Amount for Each Region:")
print(total_sales)
```

```python
import pandas as pd

cities = pd.DataFrame({
    "City_ID": ["C101", "C102", "C103", "C104", "C105"],
    "City_Name": ["Chennai", "Bangalore", "Hyderabad", "Mumbai", "Delhi"],
    "State": ["Tamil Nadu", "Karnataka", "Telangana", "Maharashtra", "Delhi"]
})

weather = pd.DataFrame({
    "City_ID": [
        "C101", "C101",
        "C102", "C102",
        "C103", "C103",
        "C104", "C104",
        "C105", "C105"
    ],
    "Date": [
        "2026-08-01", "2026-08-02",
        "2026-08-01", "2026-08-02",
        "2026-08-01", "2026-08-02",
        "2026-08-01", "2026-08-02",
        "2026-08-01", "2026-08-02"
    ],
    "Temperature": [32, 34, 28, 30, 35, 36, 31, 33, 37, 38],
    "Humidity": [75, 70, 65, 60, 55, 50, 70, 68, 45, 42],
    "Rainfall": [10, 5, 8, 4, 2, 1, 12, 7, 0, 0]
})

merged = pd.merge(cities, weather, on="City_ID")

print("Merged Data:")
print(merged[[
    "City_Name",
    "State",
    "Date",
    "Temperature",
    "Humidity",
    "Rainfall"
]])
average_temp = merged.groupby("City_Name")["Temperature"].mean()

print("\nAverage Temperature of Each City:")
print(average_temp)
highest_city = average_temp.idxmax()
highest_temperature = average_temp.max()

print("\nCity with Highest Average Temperature:")
print(highest_city)
print("Average Temperature:", highest_temperature)
```
